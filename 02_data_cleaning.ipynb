# 🧹 Data Cleaning — Fixing Messy Real-World Data

Real data is never perfect. In this notebook, we take a dirty version of our 
dataset and systematically clean it by:
- Removing users with missing names
- Eliminating duplicate friend entries
- Dropping inactive users (no friends, no liked pages)
- Deduplicating pages with repeated IDs

**Key concepts:** List comprehensions, sets, dictionary deduplication

```python
import json
def load_data(filename):
    with open("coders2.json","r") as file:
        data=json.load(file)
    return data
data=load_data("coders2.json")
def clean_data(data):
    #Remove missing values:
    data["users"]=[user for user in data["users"] if user['name'].strip()] #len=0 cannot be used as there might be a blank space(len=1)
                                                        #strip() is used to remove spaces,white tabs etc
    #Remove duplicates:
    for user in data["users"]:
        user['friends']=list(set(user['friends'])) #created a set
    #Remove inactive users:
    data["users"]=[users for users in data["users"] if users['liked_pages'] or users['friends']] #the empty list is not considered
    #Remove duplicate pages:
    unique_pages={}
    for page in data["pages"]:
        unique_pages[page['id']]=page
    data["pages"]=list(unique_pages.values())
    return data

data=clean_data(data)
json.dump(data,open("cleaned_coders2.json","w"),indent=4)
print("Data has been succesfully cleaned")
```
