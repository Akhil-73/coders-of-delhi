# 📂 Introduction — Loading & Exploring User Data

In this notebook, we set up our CodeBook dataset by creating and loading a 
JSON file containing user profiles, their friend connections, and liked pages.
We then write a function to display this data in a readable, structured format.

**Key concepts:** JSON creation, file I/O, dictionary traversal

```python
#creating the json file
import json
data={
"users": [
{"id": 1, "name": "Amit", "friends": [2, 3], "liked_pages": [101]},
{"id": 2, "name": "Priya", "friends": [1, 4], "liked_pages": [102]},
{"id": 3, "name": "Rahul", "friends": [1], "liked_pages": [101, 103]},
{"id": 4, "name": "Sara", "friends": [2], "liked_pages": [104]}
],
"pages": [
{"id": 101, "name": "Python Developers"},
{"id": 102, "name": "Data Science Enthusiasts"},
{"id": 103, "name": "AI & ML Community"},
{"id": 104, "name": "Web Dev Hub"}]}
with open("coders.json","w") as file:
    json.dump(data,file)
print(data)
```

```python
#Loading the data
def load_data(filename):
    with open("coders.json","r")as file:
        data=json.load(file)
    return data
```

```python
data=load_data("coders.json")
print(data)
```

```python
#Write a function to display users and their connections 
def users_information(data):
    print("Users and their connection:")
    for user in data['users']:
        print(f"ID:{user['id']}- {user['name']} is freinds with {user['friends']} and liked pages are:{user['liked_pages']}")
    print("\nPages and their information:")
    for page in data['pages']:
        print(f"{page['id']}- {page['name']}")        
```

```python
users_information(data)
```
