# 📄 Pages You Might Like — Collaborative Filtering

Our final feature recommends pages to users based on shared interests with 
other users. If two users like some of the same pages, they likely share 
broader interests — so we suggest each other's pages to them.

This is a simplified but real implementation of **collaborative filtering**, 
the same concept used by Netflix, Spotify, and LinkedIn.

**Key concepts:** Set intersection, similarity scoring, sorted recommendations

```python
import json
def load_data(filename):
    with open(filename,"r") as file:
        return json.load(file)

def pages_you_might_like(user_id,data):
    #Dictionary to store the user interactions pages.
    user_pages={}
    for user in data['users']:
        user_pages[user['id']]=set(user['liked_pages'])
    #If user is not found then return empty list
    if user_id not in user_pages:
        return[]
    #For pages which user directly likes:
    user_liked_pages=user_pages[user_id]
    page_suggestions={}

    #For user_id and other_user 
    for other_user,pages in user_pages.items(): #items returns view object of tuples containing key-value pair(it is like a list of tuples)
                                                #other_user-key and pages-value
        if other_user != user_id:  #we want other users pages as well
            shared_pages=user_liked_pages.intersection(pages) #pages that the user_id and other shares
        for page in pages: #iterating over pages
            if page not in user_liked_pages:
                page_suggestions[page]=page_suggestions.get(page,0)+len(shared_pages) #for counting to the maximum
        
    sorted_pages=sorted(page_suggestions.items(),key=lambda x:x[1],reverse=True)#to sort the page in decreasing order, and get the maximum probable page for suggesting.
                                    #page_suggestion.items() generates the tuple:(page_id,score), then x is for each tuple,
                                    #meaning at first, x=(104,2) then x[1] is 2, so it is sorted using scores
    return [page_id for page_id,_ in sorted_pages] #"__" is a throwaway variable, by convention, python ignores the value of the variable 
            
data=load_data("coders2.json")
user_id=int(input("Enter the valid user_id"))
page_recommendation=pages_you_might_like(user_id,data)
print(page_recommendation)
```
